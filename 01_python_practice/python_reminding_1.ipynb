# 파이썬 기본 개념 복습 노트북

파이썬의 핵심 개념을 셀 단위로 실행하면서 복습할 수 있도록 구성한 자료
설명 + 예제 코드로 구성. 기억이 가물가물할 때 활용

## 1. 변수와 기본 자료형

변수는 값을 저장하는 이름입니다. 파이썬은 변수에 타입을 미리 적지 않아도 됩니다.
이 때문에 코드가 비교적 짧고 읽기 쉽습니다.

```python
# 변수에 값을 저장합니다.
name = "Alice"
age = 25
height = 165.5
is_student = True

print(name, age, height, is_student)
print(type(name), type(age), type(height), type(is_student))
```

## 2. 문자열 다루기

문자열은 텍스트를 저장하는 자료형입니다. 인덱싱, 슬라이싱, 연결, 반복 등을 통해 자주 다룹니다.

```python
message = "파이썬은 재밌어요"

print(message)
print(message[0])
print(message[-1])
print(message[:3])
print(message + "!")
print(message * 2)
print(f"안녕, {name}님")
```

## 3. 연산자

연산자는 숫자를 계산하거나 값을 비교할 때 사용합니다.
기본적인 사칙연산 외에도 몫, 나머지, 제곱 연산이 있습니다.

```python
a = 10
b = 3

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** b)
print(a > b)
```

## 4. 조건문

조건문은 특정 조건이 참인지 거짓인지에 따라 다른 코드를 실행하게 합니다.
if, elif, else를 사용합니다.

```python
score = 88

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
else:
    grade = "C"

print(f"점수: {score}, 등급: {grade}")
```

## 5. 반복문

반복문은 같은 작업을 여러 번 실행할 때 사용합니다.
for는 반복 가능한 객체를 순회할 때, while은 조건이 참인 동안 계속 실행됩니다.

```python
for i in range(3):
    print(f"for 루프 {i}")

count = 0
while count < 3:
    print(f"while 루프 {count}")
    count += 1
```

## 6. 리스트, 튜플, 딕셔너리, 집합

컬렉션은 여러 데이터를 한 번에 저장할 때 사용합니다.
각 자료형의 특징을 이해하면 데이터를 다루기 훨씬 쉬워집니다.

```python
fruits = ["apple", "banana", "cherry"]
fruits.append("date")
print("리스트:", fruits)

coordinates = (10, 20)
print("튜플:", coordinates)

person = {"name": "Bob", "age": 30}
print("딕셔너리:", person)
print(person["name"])

numbers = {1, 2, 2, 3, 3, 4}
print("집합:", numbers)
```

## 7. 함수

함수는 특정 작업을 재사용 가능한 단위로 묶어둔 것입니다.
코드의 중복을 줄이고 구조를 명확하게 만드는 데 매우 중요합니다.

```python
def greet(name: str) -> str:
    return f"안녕하세요, {name}님"

def add(a: int, b: int) -> int:
    return a + b

print(greet("Minji"))
print(add(5, 7))
```

## 8. 람다 함수와 map/filter

람다 함수는 이름 없이 짧게 함수를 정의할 때 사용합니다.
map과 filter는 데이터를 변형하거나 조건에 맞는 요소만 걸러낼 때 자주 쓰입니다.

```python
square = lambda x: x * x
nums = [1, 2, 3, 4, 5]

print(list(map(square, nums)))
print(list(filter(lambda x: x % 2 == 0, nums)))
```

## 9. 리스트 컴프리헨션

리스트 컴프리헨션은 반복문을 더 간결하게 작성하는 문법입니다.
짧은 코드로 리스트를 만들 때 매우 유용합니다.

```python
squares = [x * x for x in range(1, 6)]
even_numbers = [x for x in range(1, 11) if x % 2 == 0]

print(squares)
print(even_numbers)
```

## 10. 예외 처리

예외 처리는 예상치 못한 오류가 발생했을 때 프로그램이 멈추지 않도록 도와줍니다.

```python
try:
    result = 10 / 0
except ZeroDivisionError as e:
    print("0으로 나눌 수 없습니다:", e)
finally:
    print("이 부분은 항상 실행됩니다")
```

## 11. 파일 입출력

파일에 데이터를 쓰거나 읽을 때 사용합니다.
with 구문을 쓰면 파일을 자동으로 닫아 주어 안전합니다.

```python
with open("/tmp/python_practice_sample.txt", "w", encoding="utf-8") as file:
    file.write("파이썬 파일 쓰기 테스트\n")
    file.write("이 줄은 파일에 저장됩니다")

with open("/tmp/python_practice_sample.txt", "r", encoding="utf-8") as file:
    print(file.read())
```

## 12. 클래스와 객체

클래스는 객체를 만들기 위한 설계도입니다.
객체는 클래스에 정의된 속성과 행동을 가지게 됩니다.

```python
class Dog:
    def __init__(self, name: str):
        self.name = name

    def bark(self) -> str:
        return f"{self.name}가 멍멍!"

dog = Dog("초코")
print(dog.bark())
```

## 13. 모듈 import

모듈은 재사용 가능한 코드들을 묶어둔 파일입니다.
import를 사용해 표준 라이브러리나 다른 파일의 기능을 가져올 수 있습니다.

```python
import math

print(math.sqrt(16))
print(math.pi)
```

## 14. 실전 예제

지금까지 배운 내용을 조합해서 아주 작은 프로그램을 만들어 봅니다.

```python
user_input = input("숫자를 입력하세요: ")
number = int(user_input)

if number % 2 == 0:
    print(f"{number}는 짝수입니다")
else:
    print(f"{number}는 홀수입니다")
```
